### Step 3: Exploratory Data Analysis (EDA)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/world_food_prices_cleaned.csv")
df.head(10)
```

```python
df.columns
```

###Q1. What is the distrubuation house price?

```python
plt.figure (figsize=(10,5))
sns.histplot(df['close'], bins=30)
plt.title('Distribution of food Prices')
plt.xlabel('Close')
plt.ylabel('Count')
plt.show()
```

### Analysis:-
1. Compared restaurant ratings.
2. Identified popular restaurants.

### Conclusion:-

1.Found useful food trends and helped imporve customer satisfaction.

### How does Food affect price

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df['country'], y=df['close'])

plt.title('How Living Area (food) Affects Prices')
plt.xlabel('Country')
plt.ylabel('Food Price (Close)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### analysis

1. Living area is an important factor in deteminig food price

### conclusion

1. Living area has significant impact on food prices
